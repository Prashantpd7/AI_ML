```python
def understand_data(df, name="Data"):
    print("=" * 80)
    print(f"DATASET: {name}")
    print("=" * 80)
    print(f"Rows: {df.shape[0]}")
    print(f"Columns: {df.shape[1]}")
    print("\nCOLUMN DETAILS:")
    
    details = pd.DataFrame({
        "Data Type": df.dtypes,
        "Missing": df.isnull().sum(),
        "Missing %": (df.isnull().mean() * 100).round(2),
        "Unique": df.nunique(),
        "Duplicates": df.duplicated().sum()
    })
    
    print(details)
    
    print("\nNUMERICAL SUMMARY:")
    display(df.describe().T)
    
    print("\nCATEGORICAL COLUMN VALUES:")
    for col in df.select_dtypes(include="object").columns:
        print(f"\n--- {col} ---")
        print("Unique:", df[col].nunique())
        print(df[col].value_counts().head(10))
    
    print("\nSAMPLE ROWS:")
    display(df.head(3))


# Run for all datasets
datasets = {
    "Orders": orders,
    "Customers": customers,
    "Order Items": order_items,
    "Payments": order_payments,
    "Geolocation": geolocation,
    "Reviews": order_reviews,
    "Products": products,
    "Sellers": sellers,
    "Translation": translation
}

for name, df in datasets.items():
    understand_data(df, name)
```

```python
# ============================================================
# REMAINING DATA UNDERSTANDING
# ============================================================

print("\n" + "="*60)
print("1. PRIMARY KEY CHECK")
print("="*60)

print("Customers - customer_id unique:",
      customers["customer_id"].is_unique)

print("Orders - order_id unique:",
      orders["order_id"].is_unique)

print("Order Items - order_item_id unique:",
      order_items["order_item_id"].is_unique)

print("Products - product_id unique:",
      products["product_id"].is_unique)


# ============================================================
# 2. FOREIGN KEY CHECK
# ============================================================

print("\n" + "="*60)
print("2. FOREIGN KEY CHECK")
print("="*60)

# Orders → Customers
invalid_customers = ~orders["customer_id"].isin(
    customers["customer_id"]
)

print("Orders with invalid customer_id:",
      invalid_customers.sum())


# Order Items → Orders
invalid_orders = ~order_items["order_id"].isin(
    orders["order_id"]
)

print("Order items with invalid order_id:",
      invalid_orders.sum())


# Order Items → Products
invalid_products = ~order_items["product_id"].isin(
    products["product_id"]
)

print("Order items with invalid product_id:",
      invalid_products.sum())


# ============================================================
# 3. NEGATIVE / ZERO VALUE CHECK
# ============================================================

print("\n" + "="*60)
print("3. NEGATIVE / ZERO VALUE CHECK")
print("="*60)

print("Negative price:",
      (order_items["price"] < 0).sum())

print("Zero price:",
      (order_items["price"] == 0).sum())

print("Negative freight value:",
      (order_items["freight_value"] < 0).sum())

print("Zero freight value:",
      (order_items["freight_value"] == 0).sum())


# ============================================================
# 4. DATE CONSISTENCY CHECK
# ============================================================

print("\n" + "="*60)
print("4. DATE CONSISTENCY CHECK")
print("="*60)

date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for column in date_columns:
    orders[column] = pd.to_datetime(
        orders[column],
        errors="coerce"
    )

print(
    "Approved before purchase:",
    (
        orders["order_approved_at"]
        < orders["order_purchase_timestamp"]
    ).sum()
)

print(
    "Carrier delivery before purchase:",
    (
        orders["order_delivered_carrier_date"]
        < orders["order_purchase_timestamp"]
    ).sum()
)

print(
    "Customer delivery before purchase:",
    (
        orders["order_delivered_customer_date"]
        < orders["order_purchase_timestamp"]
    ).sum()
)

print(
    "Estimated delivery before purchase:",
    (
        orders["order_estimated_delivery_date"]
        < orders["order_purchase_timestamp"]
    ).sum()
)


# ============================================================
# 5. ORDER-ITEM RELATIONSHIP
# ============================================================

print("\n" + "="*60)
print("5. ORDER-ITEM RELATIONSHIP")
print("="*60)

items_per_order = order_items.groupby("order_id").size()

print("Average items per order:",
      items_per_order.mean())

print("Maximum items in one order:",
      items_per_order.max())

print(
    "Orders without order items:",
    (~orders["order_id"].isin(order_items["order_id"])).sum()
)


# ============================================================
# 6. DUPLICATE ID CHECK
# ============================================================

print("\n" + "="*60)
print("6. DUPLICATE ID CHECK")
print("="*60)

print("Duplicate customer IDs:",
      customers["customer_id"].duplicated().sum())

print("Duplicate order IDs:",
      orders["order_id"].duplicated().sum())

print("Duplicate order item IDs:",
      order_items["order_item_id"].duplicated().sum())

print("Duplicate product IDs:",
      products["product_id"].duplicated().sum())


# ============================================================
# 7. FINAL CHECK
# ============================================================

print("\n" + "="*60)
print("DATA UNDERSTANDING CHECKS COMPLETED")
print("="*60)
```

```python
# ============================================================
# OUTLIER CHECK - REMAINING CHECKS
# ============================================================

print("=" * 60)
print("OUTLIER CHECK - REMAINING")
print("=" * 60)


# ------------------------------------------------------------
# 1. PRODUCT WEIGHT OUTLIERS
# ------------------------------------------------------------

Q1 = products["product_weight_g"].quantile(0.25)
Q3 = products["product_weight_g"].quantile(0.75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

weight_outliers = products[
    (products["product_weight_g"] < lower_limit) |
    (products["product_weight_g"] > upper_limit)
]

print("\nPRODUCT WEIGHT")
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)
print("Number of outliers:", len(weight_outliers))


# ------------------------------------------------------------
# 2. PRODUCT DIMENSION OUTLIERS
# ------------------------------------------------------------

dimension_columns = [
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]

print("\nPRODUCT DIMENSIONS")

for column in dimension_columns:

    Q1 = products[column].quantile(0.25)
    Q3 = products[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = products[
        (products[column] < lower_limit) |
        (products[column] > upper_limit)
    ]

    print("\n", column)
    print("Q1:", Q1)
    print("Q3:", Q3)
    print("IQR:", IQR)
    print("Lower limit:", lower_limit)
    print("Upper limit:", upper_limit)
    print("Number of outliers:", len(outliers))


print("\n" + "=" * 60)
print("OUTLIER CHECK COMPLETED")
print("=" * 60)
```
